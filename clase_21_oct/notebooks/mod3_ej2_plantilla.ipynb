## Algoritmo de Grover: Oráculo de Fase, Difusor y Muestreo

### 1. Contexto
El algoritmo de Grover localiza un elemento objetivo $\omega$ en una base de $N = 2^n$ elementos en $k \approx \dfrac{\pi}{4}\sqrt{N}$ iteraciones.
Consta de dos bloques unitarios:
1. **Oráculo $U_\omega$:** Invierte el signo del estado objetivo.
2. **Difusor $D$:** Invierte las amplitudes respecto a la media.

---

### 2. Referencia Rápida
| Operación | Código |
| :--- | :--- |
| Compuerta MCX | `qc.mcx(control_qubits, target_qubit)` |
| Multi-Controlled Z | `qc.h(target); qc.mcx(controls, target); qc.h(target)` |
| Ensamble modular | `qc_principal.compose(subcircuito, inplace=True)` |
