# Desafíos industriales para computación cuántica

### Ejercicio: Coordinación confiable con un estado GHZ
* **Caso industrial:** Tres nodos críticos deben validar correlación cuántica antes de usar una señal coordinada.

### Ejercicio: Costos y cuotas de ejecución cuántica
* **Caso industrial:** Un equipo debe estimar el consumo de su cuota antes de enviar experimentos a hardware cuántico.

### Ejercicio: Búsqueda de Grover en dos qubits
* **Caso industrial:** Un sistema debe localizar un código válido entre cuatro estados posibles mediante una regla de caja negra.

### Ejercicio: Escalando la base de datos: Grover con 5 qubits
* **Caso industrial:** Una aseguradora necesita localizar la póliza binaria `10110` dentro de 32 registros.

### Ejercicio: Optimización del compilador cuántico
* **Caso industrial:** Un circuito Grover de tres qubits debe adaptarse a una topología lineal de hardware.

### Ejercicio: Simulación cuántica: GPU frente a CPU
* **Caso industrial:** Un laboratorio debe elegir una plataforma para simulaciones de vector de estado de mayor escala.
